# Data Consistency Checks

### Handling Missing Values

```python
# Checking for missing values
vic_listings.isnull().sum()
```

##### Note: The neighbourhood column has 41.25% missing data, so it will be dropped as the majority of its values are missing, making it less useful for analysis. The price column has 16.45% missing data, which can be imputed using the mean based on relevant features. However, since the objective is to predict price using correlated features, removing rows with missing price values is a better approach to avoid introducing bias through imputation. It is important to note that this decision may lead to bias in the dataset, potentially impacting the accuracy of the results. Columns with less than 5% missing data will be retained since the amount of missing data is insignificant. For numerical columns with more than 5% missing data, imputation will be considered if it is necessary for the analysis. The columns bathrooms_text and bathrooms mean the same thing, so we will remove bathrooms_text. The host_response_time is not required for further analysis.

```python
# Removing neighborhood column
vic_listings = vic_listings.drop(columns = ['neighbourhood', 'host_response_time', 'bathrooms_text'])
```

```python
# Removing missing prices row
vic_listings = vic_listings[~vic_listings['price'].isnull()]
```

```python
# Removing $ and comma(,) from the price column so we can extract the actual number when we want to perform aggregation when required
vic_listings['price'] = vic_listings['price'].replace({'\$' : '' , ',' : ''}, regex = True)
```

```python
vic_listings['price'].value_counts()
```

```python
# Converting price to float
vic_listings['price'] = vic_listings['price'].astype('float')
```

```python
# Converting listing_count to float
vic_listings['listings_count'] = vic_listings['listings_count'].astype('float')
```

```python
# Removing % sign and convert acceptance rate to float
vic_listings['host_acceptance_rate'] = vic_listings['host_acceptance_rate'].replace({'\%':''},regex = True).astype('float')
```

```python
# Adding mean values to host_acceptance_rate
vic_listings.loc[vic_listings['host_acceptance_rate'].isnull(),'host_acceptance_rate'] = vic_listings['host_acceptance_rate'].mean()
```

```python
# Removing % sign and convert response rate to float
vic_listings['host_response_rate'] = vic_listings['host_response_rate'].replace({'\%':''},regex = True).astype('float')
```

```python
# Adding mean values to host_response_rate
vic_listings.loc[vic_listings['host_response_rate'].isnull(),'host_response_rate'] = vic_listings['host_response_rate'].mean()
```

```python
# Adding mean values to review_scores_value
vic_listings.loc[vic_listings['review_scores_value'].isnull(),'review_scores_value'] = vic_listings['review_scores_value'].mean()
```
